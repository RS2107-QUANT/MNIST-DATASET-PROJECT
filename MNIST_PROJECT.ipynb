```python
import torch
import torch.nn as nn
import pandas as pd
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
```

```python
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784', version=1, as_frame=True)

X = mnist.data
y = mnist.target
y=y.astype(int)
X=X/255
```

```python
class ANN(nn.Module):
    def __init__(self):
        super().__init__()

        self.network = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, 10)
        )

    def forward(self, x):
        return self.network(x)

model=ANN()
```

```python
optimizer=torch.optim.Adam(model.parameters(),lr=0.001)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.142857)
```

```python
X_train
```

```python
y_train
```

```python
X_train_tensor=torch.tensor(X_train.values,dtype=torch.float32)
y_train_tensor=torch.tensor(y_train.values,dtype=torch.int64)
X_test_tensor=torch.tensor(X_test.values,dtype=torch.float32)
y_test_tensor=torch.tensor(y_test.values,dtype=torch.int64)
```

```python
y_pred_train=model(X_train_tensor)
y_pred_test=model(X_test_tensor)
```

```python
loss_func=nn.CrossEntropyLoss()
loss_train=loss_func(y_pred_train,y_train_tensor)
print(loss_train)
```

```python
for epoch in range(25):
  optimizer.zero_grad()
  y_pred_train=model(X_train_tensor)
  loss_train=loss_func(y_pred_train,y_train_tensor)
  loss_train.backward()
  optimizer.step()
  print(f"Epoch {epoch+1}: Loss {loss_train.item()}")
```

```python
y_test_pred=model(X_test_tensor)
loss_test=loss_func(y_test_pred,y_test_tensor)
print(loss_test)
```

```python
from sklearn.linear_model import LogisticRegression
log_reg=LogisticRegression(max_iter=100)
log_reg.fit(X_train_tensor,y_train_tensor)
```

```python
y_pred_log=log_reg.predict(X_test_tensor)
```

```python
log_accuracy = accuracy_score(
    y_test_tensor,
    y_pred_log
)

log_precision = precision_score(y_test_tensor,y_pred_log,average="weighted",zero_division=0)
log_recall = recall_score(y_test_tensor,y_pred_log,average="weighted",zero_division=0)
log_f1 = f1_score(y_test_tensor,y_pred_log,average="weighted",zero_division=0)
print("Accuracy:", log_accuracy)
print("Precision:", log_precision)
print("Recall:", log_recall)
print("F1 Score:", log_f1)
```

```python
from sklearn.ensemble import RandomForestClassifier
clf_model=RandomForestClassifier(n_estimators=100)
clf_model.fit(X_train_tensor,y_train_tensor)
```

```python
clf_model_pred=clf_model.predict(X_test_tensor)
```

```python
rf_accuracy = accuracy_score(y_test,clf_model_pred)
rf_precision = precision_score(y_test_tensor, clf_model_pred,average="weighted",zero_division=0)
rf_recall = recall_score(y_test_tensor,clf_model_pred,average="weighted",zero_division=0)
rf_f1 = f1_score(y_test_tensor,clf_model_pred,average="weighted",zero_division=0)
print("Accuracy:", rf_accuracy)
print("Precision:", rf_precision)
print("Recall:", rf_recall)
print("F1 Score:", rf_f1)
```
